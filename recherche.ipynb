# Moteur de recherche

Ce carnet contient uniquement l’interface du moteur. Il recharge le corpus sauvegardé dans `corpus.csv` et ne relance pas les collectes Reddit ou Arxiv.

## 1. Charger le corpus existant

Les classes utilisées restent dans leurs modules Python. `Document` est importé avec `Corpus` et `SearchEngine` pour être disponible dans le carnet.

```python
from Corpus import Corpus
from Document import Document
from SearchEngine import SearchEngine

corpus = Corpus(name="tennis")
corpus.load("corpus.csv")

print(f"Corpus chargé : {len(corpus.documents)} documents")
```

## 2. Instancier le moteur

La construction de `SearchEngine` prépare les matrices TF et TF-IDF du corpus.

```python
moteur = SearchEngine(corpus)
```

## 3. Tester quelques requêtes

Chaque recherche affiche une barre `tqdm` qui indique la progression du parcours des documents.

```python
from IPython.display import display

example_queries = ["tennis", "machine learning", "robotic"]

for requete in example_queries:
    print(f"\nRésultats pour : {requete}")
    resultats = moteur.search(requete, 5)
    display(resultats)
```

## 4. Interface graphique avec ipywidgets

Saisissez les mots-cl?s, choisissez le nombre maximal de documents, puis cliquez sur **Rechercher**. Les r?sultats s?affichent dans la zone d?di?e. La barre `tqdm` indique la progression du parcours.

```python
import ipywidgets as widgets
from IPython.display import clear_output, display

titre = widgets.Label(value="Moteur de recherche")
champ_mots_cles = widgets.Text(
    value="",
    placeholder="Ex. : tennis, machine learning",
    description="Mots cl?s",
    style={"description_width": "initial"},
)
nombre_articles = widgets.IntSlider(
    value=5,
    min=1,
    max=20,
    step=1,
    description="Nombre d?articles ? extraire",
    style={"description_width": "initial"},
)
bouton_rechercher = widgets.Button(
    description="Rechercher",
    button_style="primary",
)
zone_resultats = widgets.Output()

def clique_bouton(_):
    with zone_resultats:
        clear_output(wait=True)
        resultats = moteur.search(
            champ_mots_cles.value,
            nombre_articles.value,
        )
        if resultats.empty:
            print("Aucun document pertinent trouv?.")
        else:
            display(resultats)

bouton_rechercher.on_click(clique_bouton)
interface = widgets.VBox([
    titre,
    champ_mots_cles,
    nombre_articles,
    bouton_rechercher,
    zone_resultats,
])
display(interface)
```
